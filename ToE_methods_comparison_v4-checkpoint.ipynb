# Time of Emergence of SAI's climate signals — A Methods Framework
## GMSAT  |  G6-1.5K-HiLLA vs SSP2-4.5  |  MIROC-ES2H

Central ToE definition (after Tebaldi & Friedlingstein 2013; Marotzke 2018; Samset et al. 2020):
*"In what year would there first be sufficient evidence to reject a null hypothesis that SAI has
no impact on a given variable?"* — operationalised on member-pair difference series, with
per-pair ToEs aggregated to a median or 66th percentile. Methods (e) and (f) address a related
but distinct, retrospective interpretation.

| | Method | Question asked | Counterfactual / information set | Test | Uncertainty propagated into ToE spread | Stance |
|---|---|---|---|---|---|---|
| (a) | Paired difference, same model | By year Y, can "SAI has no impact" on the level be rejected? | Paired SSP2-4.5 member, same model | One-sided one-sample t on d(2035…Y), AR(1)-adjusted n_eff | IV of two single realizations (variance-doubling under independence; less if branch-correlated — see stationarity diagnostic) | Real-time, sequential |
| (b) | Cross-model pairs, trend test | As (a), admitting the true counterfactual model is unknown; asked of the divergence *trend* | SSP2-4.5 member from a different model; level offsets irrelevant by construction | TF13 expanding-window GLS-AR(1) trend, one-sided, sustained | IV of two single realizations (σ²_A + σ²_B, independent by construction) + structural forced-response spread; warming-rate spread enters the null — see placebo test | Real-time, sequential |
| (c) | ENSO-adjusted | As (a), for an observer who also knows the ENSO state and its mean GMSAT imprint | As (a); both series ENSO-regressed before differencing | As (a), on adjusted d | As (a), with the ENSO-congruent component of IV removed from both realizations | Real-time, sequential |
| (d) | Purely empirical | With only observed GMSAT to Y and the 2035 start date, when is a departure detectable? | Extrapolated pre-2035 trend + AR(1) prediction interval — no model counterfactual | K consecutive years below the PI | IV of one realization + counterfactual error (trend-extrapolation and AR(1)-parameter uncertainty) in place of a second noise draw | Real-time |
| (e) | Signal-to-noise | When is the SAI state expected outside internal variability about the background state? | Ensemble means of both scenarios; pooled σ | S > k·σ (distance criterion, no set error rate) | None — single deterministic ToE. IV sets the threshold (σ) but is suppressed ~√10-fold per side in the signal by ensemble averaging | Retrospective |
| (f) | Li et al. analytic T_SIG | At what lead time is the paired-difference trend significant, given the realized record? | Full record of each pair; b̂, s, r₁ estimated once | Analytic solve of b̂·n > e(n), one-sided | As (a): IV of two single realizations, entering through the realized b̂, s, r₁ of each pair | Retrospective, analytic |

**Footnotes.**
1. Per-pair ToEs aggregate to a median or 66th percentile across the pair population; (e) alone yields no distribution (its sensitivity comes only from re-estimating σ and the ensemble means on member subsets).
2. (a), (c), (d), (e) test the level/state; (b), (f) test the trend — trend tests credit none of the ramp-phase level shift and are expected systematically later (see intercept diagnostic under (f)).
3. Variance-doubling assumes independent realizations; where SAI members branch from their paired background members near 2035, the difference noise starts below 2σ² and grows as the runs decorrelate (stationarity diagnostic). Cross-model pairs in (b) are independent by construction, so no such suppression applies.
4. (b)'s null admits inter-model warming-rate differences as well as SAI impact; its structural false-positive rate is quantified by the SSP-vs-SSP placebo population.
5. (d)'s prediction interval propagates trend-parameter uncertainty (implemented in this version), which dominates counterfactual error at long horizons.

**Noise definition note:** all internal-variability estimates derive from member deviations around
the relevant ensemble mean; each method's markdown states its own estimation window and pooling.

```python
def open_nc(path):
    with fsspec.open(path, mode='rb') as f:
        return xr.open_dataset(f, engine='h5netcdf', chunks={}).load()


def global_annual_mean(ds, var='tas', lat_name='lat'):
    """Area-weighted global annual mean; returns DataArray indexed by integer year."""
    da = ds[var]
    if float(da[lat_name][0]) > float(da[lat_name][-1]):
        da = da.sortby(lat_name)
    weights = np.cos(np.deg2rad(da[lat_name]))
    weights /= weights.sum()
    gmean = (da * weights).sum(lat_name).mean('lon')
    annual = weighted_annual_resample(gmean.to_dataset(name=var), var=var)[var]
    return annual.assign_coords(time=annual.time.dt.year.values)


print('Loading MIROC-ES2H...')
miroc_HiLLA_ds  = [open_nc(f) for f in miroc_HiLLA_files]
miroc_ssp245_ds = [open_nc(f) for f in miroc_ssp245_files]
print(f'  HiLLA:  {N_MEM} members, '
      f'{int(miroc_HiLLA_ds[0].time.dt.year.min())}–{int(miroc_HiLLA_ds[0].time.dt.year.max())}')
print(f'  SSP245: {N_MEM} members, '
      f'{int(miroc_ssp245_ds[0].time.dt.year.min())}–{int(miroc_ssp245_ds[0].time.dt.year.max())}')
```

## Diagnostic — is the paired-difference noise stationary?
Tebaldi & Friedlingstein (2013): the difference of two runs is not the sum of two independent
noises, "because they may not be mutually independent." If HiLLA members branch from their paired
SSP2-4.5 members near 2035, member deviations of d(t) are suppressed at short lead times and grow
as the runs decorrelate — non-stationary noise that (f) (single pooled s over the fit window) and
(a)'s fixed pooled r₁ treat as fixed.

**Rising sd** ⇒ branch-point correlation is real: matched pairing is genuinely variance-reducing
(helps (a), (c), (f); footnote 3), early-window tests are conservative. **Flat sd** ⇒ members are
effectively independent draws; pairing is bookkeeping and the difference noise ≈ sum of the two
variances — and (b)'s fallback pairs are then statistically equivalent to matched pairs.

```python
# ── Diagnostic: time evolution of paired-difference noise ─────────────────────
_common_d = np.arange(SAI_START, PLOT_END + 1)
for ts in miroc_HiLLA_anom + miroc_ssp245_anom:
    _common_d = np.intersect1d(_common_d, ts.time.values.astype(int))
d_all = np.stack([miroc_ssp245_anom[i].sel(time=_common_d).values
                  - miroc_HiLLA_anom[i].sel(time=_common_d).values
                  for i in range(N_MEM)])            # (n_members, n_years), positive = cooling
d_dev = d_all - d_all.mean(axis=0)                   # deviations around common forced divergence

print('Paired-difference noise vs lead time (member deviations around ensemble-mean d):')
print(f'{"window":>12} {"sd (K)":>8} {"lag-1 r":>8}')
for y0 in range(SAI_START, int(_common_d.max()) - 8, 10):
    y1 = min(y0 + 9, int(_common_d.max()))
    m = (_common_d >= y0) & (_common_d <= y1)
    sd  = d_dev[:, m].std(ddof=1)
    r1s = [np.corrcoef(row[:-1], row[1:])[0, 1] for row in d_dev[:, m]]
    print(f'{y0}–{y1:<6} {sd:8.3f} {np.mean(r1s):8.2f}')

W = 11
roll_yrs = _common_d[W // 2 : len(_common_d) - W // 2]
roll_sd  = np.array([d_dev[:, j:j + W].std(ddof=1)
                     for j in range(len(_common_d) - W + 1)])

fig_d, ax_d = plt.subplots(figsize=(7, 4))
ax_d.plot(roll_yrs, roll_sd, color='k', lw=2)
ax_d.set_xlabel('Year'); ax_d.set_ylabel(f'sd of paired diff ({W}-yr rolling, K)')
ax_d.set_title('Stationarity check: paired-difference noise vs lead time', fontsize='medium')
ax_d.grid(ls='--', alpha=0.3)
for sp in ('top', 'right'): ax_d.spines[sp].set_visible(False)
plt.show()
```

## Method (a) — Paired difference, same model: cumulative one-sample t-test
For each MIROC-ES2H member pair i, form d(t) = SSP2-4.5ᵢ − HiLLAᵢ (positive = cooling) and, for
each year Y, test H₀: mean d(2035…Y) = 0 against mean > 0 with a one-sample t-test on the growing
sample. This tests the difference values directly (the TF13 operationalisation in the framework
text), replacing the earlier two-sample Welch construction — the paired form also inherits any
branch-point variance suppression in d (see stationarity diagnostic).

**Serial-correlation adjustment:** effective sample size n_eff = n·(1−r₁)/(1+r₁) (Zwiers & von
Storch 1995) replaces n in the standard error and DOF. r₁ is pooled once: the ensemble-mean lag-1
autocorrelation of member deviations of d around the ensemble-mean divergence, years ≥ 2035,
clipped at 0, and held fixed (including in bootstrap subsets). No test until n_eff > 2.

Member ToE = first Y with p < α (α = 0.05 headline). Aggregation: **66th percentile** across the
10 pairs (sensitivity variants pair α with a percentile, as in the bootstrap cell).

```python
def pooled_r1_of_diff(hilla_list, ssp245_list, start_year=SAI_START, end_year=PLOT_END):
    '''Ensemble-mean lag-1 autocorrelation of member deviations of d = SSP245 − HiLLA
    around the ensemble-mean divergence.'''
    common = np.arange(start_year, end_year + 1)
    for ts in hilla_list + ssp245_list:
        common = np.intersect1d(common, ts.time.values.astype(int))
    d = np.stack([ssp245_list[i].sel(time=common).values - hilla_list[i].sel(time=common).values
                  for i in range(len(hilla_list))])
    dev = d - d.mean(axis=0)
    r1s = [np.corrcoef(row[:-1], row[1:])[0, 1] for row in dev]
    return float(np.mean(r1s))


def eff_n(n, r1):
    '''Zwiers & von Storch (1995) effective sample size; only shrink for r1 > 0.'''
    r1 = min(max(r1, 0.0), 0.99)
    return n * (1 - r1) / (1 + r1)


def paired_ttest_toe(hilla_ts, ssp245_ts, r1_d,
                     start_year=SAI_START, alpha=ALPHA, min_n=MIN_N):
    '''One-sided one-sample t-test on d = SSP245 − HiLLA (H0: mean = 0 vs mean > 0),
    on the growing sample 2035…Y, with AR(1)-adjusted effective n.

    Returns (toe, {year: p}).'''
    common = np.intersect1d(hilla_ts.time.values.astype(int),
                            ssp245_ts.time.values.astype(int))
    common = common[common >= start_year]
    d = ssp245_ts.sel(time=common).values - hilla_ts.sel(time=common).values
    p_vals = {}; toe = None
    for j, Y in enumerate(common):
        n = j + 1
        ne = eff_n(n, r1_d)
        if n < min_n or ne <= 2:
            p_vals[int(Y)] = np.nan; continue
        dw = d[:n]
        se = dw.std(ddof=1) / np.sqrt(ne)
        t_stat = dw.mean() / se
        p = float(stats.t.sf(t_stat, df=ne - 1))
        p_vals[int(Y)] = p
        if toe is None and p < alpha:
            toe = int(Y)
    return toe, p_vals


R1_D = pooled_r1_of_diff(miroc_HiLLA_anom, miroc_ssp245_anom)
print(f'Pooled lag-1 autocorrelation of paired-difference deviations (≥{SAI_START}): '
      f'r1 = {R1_D:+.3f}  (clipped at 0 in use)')

a_mem_toes = []; a_pvals_all = []
for h, s in zip(miroc_HiLLA_anom, miroc_ssp245_anom):
    toe, pvs = paired_ttest_toe(h, s, R1_D)
    a_mem_toes.append(toe); a_pvals_all.append(pvs)

a_p66_toe = int(np.percentile([t for t in a_mem_toes if t is not None], 66))
print(f'Method (a) member ToEs   : {a_mem_toes}')
print(f'Method (a) 66th pct ToE  : {a_p66_toe}')
```

## Method (b) — Cross-model pairs: TF13 GLS expanding-window trend test
Accounts for uncertainty in the counterfactual by pairing each MIROC HiLLA member with an
SSP2-4.5 member from a **different model** (Tebaldi & Friedlingstein 2013, *PNAS*,
doi:10.1073/pnas.1300005110). The test is on the **trend** of d(t) = SSP2-4.5(other) − HiLLA(MIROC):
a trend statistic ignores the intercept, so inter-model absolute-temperature offsets are
irrelevant by construction and no alignment step is needed.

- Each year Y ≥ 2035 + B_MIN_N − 1: GLS-AR(1) linear-trend fit to d(2035…Y)
  (Prais–Winsten iteration — TF13's "recursive estimation"); one-sided test (trend > 0), p < α.
- **ToE = first year significant and sustained through B_END** (TF13's rule; uses future
  information, so this is a retrospective certification, slightly earlier than strict real-time).
- Pair population: one counterfactual member per HiLLA member (10 pairs, comparable to (a));
  the full cross-product is a straightforward extension. Median across pairs.

**Null-hypothesis caveat (framework footnote 4):** under H₀ a cross-model d(t) still trends
wherever the two models warm at different rates under SSP2-4.5 — (b) detects "SAI impact OR
inter-model warming-rate difference". The **placebo test** below (other-model SSP2-4.5 vs MIROC
SSP2-4.5, no SAI anywhere) quantifies this structural false-positive rate directly.

**Data note:** with `CROSS_SSP245_FILES` empty this cell runs in FALLBACK mode — same-model
MIROC pairs with mismatched member numbering (rᵢ HiLLA vs rᵢ₊₁ SSP2-4.5). That exercises the full
pipeline with independent realizations (no branch correlation) but samples **no structural
uncertainty**, and the placebo rate should then sit near α. Replace with real cross-model paths
before quoting (b).

```python
def prais_winsten_trend(y, tol=1e-8, maxiter=100):
    '''GLS linear-trend fit with AR(1) residuals via Prais–Winsten iteration
    (TF13's 'recursive estimation').

    Returns (slope, se_slope, rho, dof, intercept); slope per time step.'''
    n = len(y)
    t_ax = np.arange(1, n + 1, dtype=float)
    X = np.column_stack([np.ones(n), t_ax])
    beta = np.linalg.lstsq(X, y, rcond=None)[0]
    rho = 0.0
    ys, Xs = y, X
    for _ in range(maxiter):
        resid = y - X @ beta                            # rho from ORIGINAL-scale residuals
        denom = float(resid[:-1] @ resid[:-1])
        rho_new = float(resid[1:] @ resid[:-1]) / denom if denom > 0 else 0.0
        rho_new = min(max(rho_new, -0.99), 0.99)
        converged = abs(rho_new - rho) < tol
        rho = rho_new
        c = np.sqrt(1 - rho ** 2)                       # Prais–Winsten: keep first obs
        ys = np.concatenate([[c * y[0]], y[1:] - rho * y[:-1]])
        Xs = np.vstack([c * X[:1],       X[1:] - rho * X[:-1]])
        beta = np.linalg.lstsq(Xs, ys, rcond=None)[0]
        if converged:
            break
    resid_s = ys - Xs @ beta
    dof = n - 2
    s2  = float(resid_s @ resid_s) / dof
    cov = s2 * np.linalg.inv(Xs.T @ Xs)
    return float(beta[1]), float(np.sqrt(cov[1, 1])), rho, dof, float(beta[0])


def tf_gls_pvals(sai_ts, cf_ts, start_year=SAI_START, end_year=None, min_n=B_MIN_N):
    '''One-sided p-value (trend in d = counterfactual − SAI > 0) for each expanding
    window [start_year, Y].'''
    if end_year is None:
        end_year = B_END
    common = np.intersect1d(sai_ts.time.values.astype(int),
                            cf_ts.time.values.astype(int))
    common = common[(common >= start_year) & (common <= end_year)]
    d = cf_ts.sel(time=common).values - sai_ts.sel(time=common).values
    p_vals = {}
    for j, Y in enumerate(common):
        n = j + 1
        if n < min_n:
            p_vals[int(Y)] = np.nan
            continue
        slope, se, rho, dof, _ = prais_winsten_trend(d[:n])
        p_vals[int(Y)] = float(stats.t.sf(slope / se, df=dof)) if se > 0 else np.nan
    return p_vals


def tf_toe_from_pvals(p_vals, alpha=ALPHA):
    '''First year Y with p < alpha in Y and ALL subsequent years (sustained through record end).'''
    yrs = sorted(p_vals)
    toe = None
    for y in reversed(yrs):
        p = p_vals[y]
        if np.isfinite(p) and p < alpha:
            toe = y
        else:
            break
    return toe


# ── counterfactual population ─────────────────────────────────────────────────
if CROSS_SSP245_FILES:
    print(f'Loading {CROSS_MODEL_NAME} SSP2-4.5 counterfactual members...')
    _cross_ds    = [open_nc(f) for f in CROSS_SSP245_FILES]
    _cross_gmsat = [global_annual_mean(ds) for ds in _cross_ds]
    _cross_ref   = float(np.mean([float(ts.sel(time=slice(*ANOM_REF)).mean())
                                  for ts in _cross_gmsat]))
    b_cf_anom = [ts - _cross_ref for ts in _cross_gmsat]   # anomalies vs own model ref
    B_FALLBACK = False
    B_LABEL = CROSS_MODEL_NAME
else:
    # FALLBACK: same-model mismatched-member pairs (independent realizations, no structure)
    b_cf_anom = [miroc_ssp245_anom[(i + 1) % N_MEM] for i in range(N_MEM)]
    B_FALLBACK = True
    B_LABEL = 'MIROC mismatched-member FALLBACK'
    print('Method (b): CROSS_SSP245_FILES empty — running same-model mismatched-member FALLBACK.')

b_pairs = [(miroc_HiLLA_anom[i], b_cf_anom[i % len(b_cf_anom)]) for i in range(N_MEM)]

print(f'Method (b) — TF13 GLS-AR(1) expanding-window trend on cross pairs '
      f'({B_LABEL}; 2035–{B_END}, one-sided, sustained):')
b_pvals_all = [tf_gls_pvals(h, cf) for h, cf in b_pairs]
b_mem_toes  = [tf_toe_from_pvals(pv) for pv in b_pvals_all]
for i, toe in enumerate(b_mem_toes):
    print(f'  pair {i + 1:02d}: ToE = {toe}')
b_median_toe = int(np.median([t for t in b_mem_toes if t is not None]))
print(f'Method (b) pair ToEs   : {b_mem_toes}')
print(f'Method (b) median ToE  : {b_median_toe}')

# ── placebo population: counterfactual vs MIROC SSP2-4.5 (no SAI anywhere) ────
placebo_pairs = [(miroc_ssp245_anom[i], b_cf_anom[(i + 1) % len(b_cf_anom)])
                 for i in range(N_MEM)]
placebo_toes = [tf_toe_from_pvals(tf_gls_pvals(s, cf)) for s, cf in placebo_pairs]
n_fp = sum(t is not None for t in placebo_toes)
print(f'Placebo (no-SAI) pairs: {placebo_toes}')
print(f"Placebo 'emergence' rate by {B_END}: {n_fp}/{len(placebo_toes)} "
      f'— structural false-positive rate for (b) at α = {ALPHA}'
      + (' (fallback mode: expect ≈ α)' if B_FALLBACK else ''))
```

### SI — same-model matched-pair trend variant (estimand bridge)
The identical TF13 trend test on the **matched same-model pairs of (a)** isolates the estimand
effect (level vs trend) from the counterfactual source: (a) vs this row differ only in the test;
this row vs (b) differ only in the pair population. Trend tests exclude the ramp-phase level
shift, so expect this later than (a) (footnote 2).

```python
si_tf_pvals = [tf_gls_pvals(h, s) for h, s in zip(miroc_HiLLA_anom, miroc_ssp245_anom)]
si_tf_toes  = [tf_toe_from_pvals(pv) for pv in si_tf_pvals]
si_tf_median = int(np.median([t for t in si_tf_toes if t is not None]))
print(f'SI same-model matched-pair TF13 trend — member ToEs: {si_tf_toes}')
print(f'SI same-model matched-pair TF13 trend — median ToE : {si_tf_median}')
print(f"Estimand gap (this − method (a) 66th pct is not like-for-like; compare medians): "
      f"median here {si_tf_median} vs (a) median "
      f"{int(np.median([t for t in a_mem_toes if t is not None]))}")
```

## Method (c) — ENSO-adjusted paired difference
As (a), but for an observer who also knows the ENSO state and its mean GMSAT imprint
(Foster & Rahmstorf 2011; Thompson et al. 2009): the ENSO-congruent component of GMSAT is
regressed out of **both** members of each pair before differencing and testing.

**Index:** *relative* Niño 3.4 — area-mean tas over 5°S–5°N, 170°–120°W **minus** the tropical
mean (20°S–20°N) — computed monthly from each member's own fields. The relative form
(van Oldenborgh et al. 2021) self-corrects for uniform forced changes, so the Niño 3.4 region's
own forced cooling under SAI is not subtracted from the signal (the windowed-detrend alternative
is noted but not used). Anomalies are relative to the monthly climatology of the paired SSP2-4.5
member's pre-2035 record (HiLLA has no pre-2035 record of its own).

**Lag:** ENSO leads GMSAT by ~3 months, embedded by averaging the index over **Oct(y−1)–Sep(y)**
against calendar-year GMSAT (partial first year permitted, ≥ ENSO_MIN_MONTHS months).

**β:** per member, OLS of linearly detrended annual GMSAT anomalies on the annual index over the
member's full pre-2035 SSP2-4.5 record; the same βᵢ is applied to both members of pair i, each
with its own contemporaneous index: tas_adj = tas − βᵢ·N34relᵢ(t). Applying the pre-2035 β under
SAI assumes the ENSO–GMSAT coupling survives deployment unchanged (stated assumption). tas is
used as an SST proxy over the index regions.

**Test:** identical to (a) — one-sided one-sample t on adjusted d, with its own pooled r₁.

```python
def region_monthly_mean(ds, lat_bounds, lon_bounds=None, var='tas'):
    '''Area-weighted monthly mean of var over a lat (and optional lon) box.
    Handles descending latitudes, −180–180 longitude conventions, and coarse grids
    (cells whose centres fall within half a grid spacing of the box are included,
    so a narrow box like Niño 3.4 never selects zero cells).'''
    da = ds[var]
    if float(da.lat[0]) > float(da.lat[-1]):
        da = da.sortby('lat')
    dlat = float(np.median(np.diff(da.lat))) if da.lat.size > 1 else 0.0
    da = da.sel(lat=slice(lat_bounds[0] - dlat / 2, lat_bounds[1] + dlat / 2))
    if lon_bounds is not None:
        lo, hi = lon_bounds
        if float(da.lon.min()) < -1.0:                      # −180–180 convention
            lo = ((lo + 180) % 360) - 180
            hi = ((hi + 180) % 360) - 180
        da = da.sortby('lon')
        dlon = float(np.median(np.diff(da.lon))) if da.lon.size > 1 else 0.0
        da = da.sel(lon=slice(lo - dlon / 2, hi + dlon / 2))
    if da.lat.size == 0 or da.lon.size == 0:
        raise ValueError('region_monthly_mean: empty selection — check box bounds vs grid')
    w = np.cos(np.deg2rad(da.lat))
    return da.weighted(w).mean(('lat', 'lon'))


def rel_n34_annual(ds, clim_source_ds=None, min_months=ENSO_MIN_MONTHS):
    '''Relative Niño 3.4 index, Oct(y−1)–Sep(y) annual means.

    rel(t) = N34(t) − tropical mean(t), monthly; anomalised against the monthly
    climatology of clim_source_ds (pre-2035 months; defaults to ds itself);
    then averaged over Oct–Sep windows labelled by the Jan–Sep year.

    Returns dict {year: index value}.'''
    rel = region_monthly_mean(ds, N34_LAT, N34_LON) - region_monthly_mean(ds, TROP_LAT)
    if clim_source_ds is None:
        clim_src = rel
    else:
        clim_src = (region_monthly_mean(clim_source_ds, N34_LAT, N34_LON)
                    - region_monthly_mean(clim_source_ds, TROP_LAT))
    clim_pre = clim_src.where(clim_src.time.dt.year < SAI_START, drop=True)
    clim = clim_pre.groupby(clim_pre.time.dt.month).mean()
    anom = rel.groupby(rel.time.dt.month) - clim

    yrs    = anom.time.dt.year.values
    months = anom.time.dt.month.values
    vals   = anom.values
    idx_year = np.where(months >= 10, yrs + 1, yrs)         # Oct–Dec count toward next year
    out = {}
    for y in np.unique(idx_year):
        sel = vals[idx_year == y]
        if len(sel) >= min_months:
            out[int(y)] = float(np.mean(sel))
    return out


def enso_beta(gmsat_anom_ts, index_dict, end_year=SAI_START - 1):
    '''OLS slope of detrended annual GMSAT anomalies on the annual index, pre-2035.'''
    yrs = np.array(sorted(y for y in index_dict if y <= end_year))
    yrs = np.intersect1d(yrs, gmsat_anom_ts.time.values.astype(int))
    g   = gmsat_anom_ts.sel(time=yrs).values
    g   = g - np.polyval(np.polyfit(yrs.astype(float), g, 1), yrs.astype(float))
    n34 = np.array([index_dict[int(y)] for y in yrs])
    slope, intercept, r, p, se = stats.linregress(n34, g)
    return float(slope), float(r), len(yrs)


def enso_adjust(gmsat_anom_ts, index_dict, beta):
    '''tas_adj = tas − β·N34rel(t), over years where the index exists.'''
    yrs = np.intersect1d(np.array(sorted(index_dict)),
                         gmsat_anom_ts.time.values.astype(int))
    adj = gmsat_anom_ts.sel(time=yrs) - beta * np.array([index_dict[int(y)] for y in yrs])
    return adj


print('Computing relative Niño 3.4 indices (monthly fields already in memory)...')
ssp_n34   = [rel_n34_annual(ds) for ds in miroc_ssp245_ds]
hilla_n34 = [rel_n34_annual(ds, clim_source_ds=miroc_ssp245_ds[i])
             for i, ds in enumerate(miroc_HiLLA_ds)]

print(f'{"mem":>4} {"beta (K/K)":>11} {"r":>6} {"n_fit":>6}')
enso_betas = []
for i in range(N_MEM):
    b_i, r_i, n_i = enso_beta(miroc_ssp245_anom[i], ssp_n34[i])
    enso_betas.append(b_i)
    print(f'r{i + 1:02d}  {b_i:+11.3f} {r_i:+6.2f} {n_i:6d}')

miroc_ssp245_enso  = [enso_adjust(miroc_ssp245_anom[i], ssp_n34[i],   enso_betas[i])
                      for i in range(N_MEM)]
miroc_HiLLA_enso   = [enso_adjust(miroc_HiLLA_anom[i],  hilla_n34[i], enso_betas[i])
                      for i in range(N_MEM)]

# variance of d removed by the adjustment (post-2035 member deviations)
def _dev_var(hl, sl):
    common = np.arange(SAI_START, PLOT_END + 1)
    for ts in hl + sl:
        common = np.intersect1d(common, ts.time.values.astype(int))
    d = np.stack([sl[i].sel(time=common).values - hl[i].sel(time=common).values
                  for i in range(len(hl))])
    return float((d - d.mean(axis=0)).var(ddof=1))

var_raw, var_adj = _dev_var(miroc_HiLLA_anom, miroc_ssp245_anom), \
                   _dev_var(miroc_HiLLA_enso, miroc_ssp245_enso)
print(f'Variance of paired-difference deviations: raw {var_raw:.4f} → adjusted {var_adj:.4f} K² '
      f'({100 * (1 - var_adj / var_raw):.0f}% removed)')

R1_D_ENSO = pooled_r1_of_diff(miroc_HiLLA_enso, miroc_ssp245_enso)
print(f'Pooled r1 of adjusted difference deviations: {R1_D_ENSO:+.3f}')

c_mem_toes = []; c_pvals_all = []
for h, s in zip(miroc_HiLLA_enso, miroc_ssp245_enso):
    toe, pvs = paired_ttest_toe(h, s, R1_D_ENSO)
    c_mem_toes.append(toe); c_pvals_all.append(pvs)

c_p66_toe = int(np.percentile([t for t in c_mem_toes if t is not None], 66))
print(f'Method (c) member ToEs   : {c_mem_toes}')
print(f'Method (c) 66th pct ToE  : {c_p66_toe}')
```

## Method (d) — Purely empirical: OLS trend + AR(1) prediction interval
The only method with **no model counterfactual**: with an 'observed' GMSAT record to year Y and
the knowledge that SAI began in 2035, when is a departure detectable? Train an OLS linear trend
on each MIROC-ES2H SSP2-4.5 member (2020–2034) as pseudo-observations, fit AR(1) to the
residuals, extrapolate the trend as the no-SAI expectation with a 95% prediction interval, and
declare ToE at the first year the paired HiLLA member sits **below the lower PI bound for K = 5
consecutive years**. Median across the 10 members.

**Parameter uncertainty (new in v4):** the PI now propagates trend-parameter uncertainty in
addition to AR(1) noise:
- **forecast variance of the fitted line**, s²·γ²·(1/n + (x − x̄)²/Sxx), which grows
  quadratically with extrapolation distance and dominates at long horizons; s is the marginal
  residual sd (n−2 dof) and γ² = (1+φ)/(1−φ) discounts the effective training sample for
  autocorrelation (Zwiers & von Storch);
- **Student-t quantile** with n−2 dof in place of the Gaussian z (finite-sample s).

Remaining known omissions, deliberate for now: φ small-sample bias correction (Marriott–Pope),
uncertainty in φ itself, and the two-sided-interval/one-sided-use asymmetry (the lower bound of a
two-sided 95% PI is an effective 97.5% one-sided test — stricter than (a)–(c) at α = 0.05).

```python
PRED_YEARS = np.arange(SAI_START, PLOT_END + 1)


def ar1_forecast(obs_ts, pred_years, ci=CI_LEVEL):
    '''OLS linear trend + AR(1) residuals h-step forecast, WITH trend-parameter
    uncertainty propagated into the prediction interval.

    Returns (mu, lower_ci, upper_ci) over pred_years.'''
    x_fit = obs_ts.time.values.astype(float)
    y_fit = obs_ts.values.astype(float)
    n_fit = len(x_fit)
    slope, intercept, *_ = stats.linregress(x_fit, y_fit)
    resid  = y_fit - (slope * x_fit + intercept)
    phi    = np.dot(resid[1:], resid[:-1]) / np.dot(resid[:-1], resid[:-1])
    innov  = resid[1:] - phi * resid[:-1]
    sigma  = np.sqrt(np.sum(innov**2) / (len(innov) - 1))    # innovation sd
    eps_last = resid[-1]

    h   = pred_years.astype(float) - x_fit[-1]
    mu  = slope * pred_years + intercept + phi**h * eps_last

    # AR(1) forecast-noise variance (as before)
    var_ar1 = sigma**2 * (1 - phi**(2 * h)) / max(1 - phi**2, 1e-8)

    # NEW: trend + intercept forecast variance, AR(1)-discounted effective sample
    s_marg   = np.sqrt(np.sum(resid**2) / (n_fit - 2))       # marginal residual sd
    gamma2   = (1 + max(phi, 0.0)) / max(1 - max(phi, 0.0), 1e-8)
    x_mean   = x_fit.mean()
    sxx      = np.sum((x_fit - x_mean)**2)
    var_trend = s_marg**2 * gamma2 * (1.0 / n_fit
                                      + (pred_years.astype(float) - x_mean)**2 / sxx)

    tq = stats.t.ppf((1 + ci) / 2, df=n_fit - 2)             # t replaces z
    pi_half = tq * np.sqrt(var_ar1 + var_trend)
    return mu, mu - pi_half, mu + pi_half


def ar1_toe_lower(hilla_ts, pred_years, lower_pi,
                  start_year=SAI_START, K=K_CONSEC):
    '''First year HiLLA is below lower PI for K consecutive years.'''
    h_yrs  = hilla_ts.time.values
    common = np.intersect1d(h_yrs[h_yrs >= start_year], pred_years)
    if len(common) < K:
        return None
    h_vals = hilla_ts.sel(time=common).values
    lo     = lower_pi[np.searchsorted(pred_years, common)]
    below  = h_vals < lo
    for i in range(len(below) - K + 1):
        if below[i: i + K].all():
            return int(common[i])
    return None


print(f'Method (d) — OLS+AR(1) per member (training 2020–{SAI_START - 1}, '
      f'parameter uncertainty in PI):')
d_mem_toes      = []
ar1_results_all = []   # (mu, lo, hi) per member

for i in range(N_MEM):
    obs_i            = miroc_ssp245_anom[i].sel(time=slice(2020, SAI_START - 1))
    mu_i, lo_i, hi_i = ar1_forecast(obs_i, PRED_YEARS)
    toe_i            = ar1_toe_lower(miroc_HiLLA_anom[i], PRED_YEARS, lo_i)
    d_mem_toes.append(toe_i)
    ar1_results_all.append((mu_i, lo_i, hi_i))
    print(f'  r{i + 1:02d}: ToE = {toe_i}')

d_median_toe = int(np.median([t for t in d_mem_toes if t is not None]))
print(f'Method (d) member ToEs : {d_mem_toes}')
print(f'Method (d) median ToE  : {d_median_toe}')
```

## Method (e) — Signal-to-noise (retrospective)
"When do we expect the SAI climate state to be outside the internal-variability range around the
background state?" Assumes knowledge of the forced response (ensemble means of both scenarios) —
a retrospective separation criterion, not a real-time significance test (no set error rate;
Hawkins & Sutton lineage; see Li et al. 2017 critique).

**Signal:** SSP2-4.5 ensemble mean − HiLLA ensemble mean at each year.
**Noise σ:** sd of SSP2-4.5 member deviations around the SSP2-4.5 ensemble mean, pooled over
2025–2044 (fixed scalar).  **ToE:** first year Signal > 1.645σ. Single deterministic ToE.

```python
def signal_to_noise_toe(hilla_list, ssp245_list,
                        noise_win=NOISE_WIN, k=SN_K,
                        plot_start=PLOT_START, plot_end=PLOT_END):
    '''
    Signal  = SSP245_ens_mean(Y) - HiLLA_ens_mean(Y)   [cooling is positive]
    Noise σ = std of SSP245 member residuals around SSP245 ens mean,
              pooled over all members and years in noise_win.
    ToE     = first year Signal > k * σ.
    '''
    noise_years = np.arange(noise_win[0], noise_win[1] + 1)
    ssp_noise_vals = np.stack([
        ts.sel(time=noise_years).values for ts in ssp245_list
    ])  # (n_members, n_noise_years)
    ssp_noise_mean = ssp_noise_vals.mean(axis=0)  # ensemble mean per year
    residuals = ssp_noise_vals - ssp_noise_mean    # member deviations
    sigma = float(residuals.std(ddof=1))           # pooled std (scalar)
    threshold = k * sigma
    print(f'Noise σ (SSP245 internal variability, {noise_win[0]}–{noise_win[1]}): {sigma:.4f} K')
    print(f'Threshold ({k} σ): {threshold:.4f} K')

    all_yrs = np.arange(SAI_START, plot_end + 1)
    common_yrs = all_yrs.copy()
    for ts in hilla_list + ssp245_list:
        common_yrs = np.intersect1d(common_yrs, ts.time.values.astype(int))

    hilla_mean = np.stack([ts.sel(time=common_yrs).values for ts in hilla_list]).mean(axis=0)
    ssp_mean   = np.stack([ts.sel(time=common_yrs).values for ts in ssp245_list]).mean(axis=0)
    signal     = ssp_mean - hilla_mean   # positive = cooling

    exceed = np.where(signal > threshold)[0]
    toe = int(common_yrs[exceed[0]]) if len(exceed) else None

    return sigma, threshold, common_yrs, signal, toe


sn_sigma, sn_threshold, sn_years, sn_signal, sn_toe = signal_to_noise_toe(
    miroc_HiLLA_anom, miroc_ssp245_anom
)
print(f'Method (e) (signal-to-noise)  ToE: {sn_toe}')
```

## Method (f) — Li et al. (2017) analytic T_SIG on paired differences (retrospective)
Closed-form lead time for a linear trend to emerge from Gaussian AR(1) noise
(Li, Thompson, Barnes & Solomon 2017, *J. Climate*, doi:10.1175/JCLI-D-16-0280.1),
derived from the standard error of the regression.

For each member pair, form d(t) = SSP2-4.5 − HiLLA (positive = cooling signal), fit OLS over
2035–F_END, and estimate b̂ (trend per year), s (residual sd) and r1 (lag-1 autocorrelation of
the residuals) **once, from the full fit window**. ToE = first n (time steps since 2035) with

$$\hat{b}\,n \; > \; t_c(n-2)\; s\, \sqrt{\tfrac{1+r_1}{1-r_1}}\, \sqrt{\tfrac{12n}{n^2-1}}$$

i.e. the total realized change exceeds the margin of error of the regression at the chosen
confidence level. Quoted as the **median ToE across the 10 member pairs**.

**Conventions (fixed choices):**
- **one-sided** t_c — cooling direction known a priori, consistent with (a)–(c) (Li et al. use two-tailed)
- **fit window ends F_END = 2065**, where the paired-difference trend is still approximately linear
- **paired member differences** — differencing removes the common forced warming, and the
  difference of two ≈AR(1) series remains ≈AR(1)

Parameters come from the full realized record, so this is a retrospective/analytic emergence
time — which is exactly why it is immune to the sequential-testing inflation affecting (a)–(d).

```python
def li_fit_diff(hilla_ts, ssp245_ts, start_year=SAI_START, end_year=F_END):
    '''OLS fit to the paired difference d(t) = SSP245 - HiLLA over [start_year, end_year].'''
    common = np.intersect1d(hilla_ts.time.values.astype(int),
                            ssp245_ts.time.values.astype(int))
    common = common[(common >= start_year) & (common <= end_year)]
    d = ssp245_ts.sel(time=common).values - hilla_ts.sel(time=common).values
    n = len(d)
    t_ax = np.arange(1, n + 1, dtype=float)              # time steps 1..n (n=1 is start_year)
    slope, intercept, *_ = stats.linregress(t_ax, d)
    resid = d - (intercept + slope * t_ax)
    s  = float(np.sqrt(np.sum(resid**2) / (n - 2)))      # regression residual sd (n-2 dof)
    r1 = float(np.corrcoef(resid[:-1], resid[1:])[0, 1]) # lag-1 autocorrelation of residuals
    gamma = float(np.sqrt((1 + r1) / max(1 - r1, 1e-8))) # AR(1) inflation, Li et al. Eq. (2)
    return {'years': common, 'd': d, 'n': n, 'slope': slope, 'intercept': intercept,
            's': s, 'r1': r1, 'gamma': gamma, 'start_year': start_year}


def li_tsig(fit, ci=F_CI, one_sided=True):
    '''Li et al. (2017) Eq. (3): smallest n with b*n > t_c(n-2) * s * gamma * g(n),
    g(n) = sqrt(12n/(n^2-1)). Solved iteratively because t_c depends on n.

    Returns (ToE year, n array, margin-of-error array e(n)) — the arrays for plotting.'''
    if fit['slope'] <= 0:
        return None, None, None
    nts = np.arange(3, fit['n'] + 1)
    q   = ci if one_sided else (1 + ci) / 2
    tc  = stats.t.ppf(q, df=nts - 2)
    e   = tc * fit['s'] * fit['gamma'] * np.sqrt(12.0 * nts / (nts**2 - 1.0))
    exceed = np.where(fit['slope'] * nts > e)[0]
    toe = int(fit['start_year'] + nts[exceed[0]] - 1) if len(exceed) else None
    return toe, nts, e


print(f'Method (f) — Li et al. T_SIG per paired member '
      f'(fit 2035–{F_END}, one-sided {int(F_CI * 100)}%):')
li_fits_all = [li_fit_diff(h, s) for h, s in zip(miroc_HiLLA_anom, miroc_ssp245_anom)]
li_toes_all = []
for i, fit in enumerate(li_fits_all):
    toe_i, _, _ = li_tsig(fit)
    li_toes_all.append(toe_i)
    print(f"  r{i + 1:02d}: b = {fit['slope']:+.4f} K/yr, s = {fit['s']:.3f} K, "
          f"r1 = {fit['r1']:+.2f}  ->  ToE = {toe_i}")

li_median_toe = int(np.median([t for t in li_toes_all if t is not None]))
print(f'Method (f) member ToEs : {li_toes_all}')
print(f'Method (f) median ToE  : {li_median_toe}')
```

### Diagnostic — how much of method (f)'s separation sits in the intercept?
The Li criterion counts only the slope contribution b·n; the fast ramp-phase cooling is a level
shift that OLS parks in the intercept, contributing nothing to emergence (footnote 2). A large
intercept share (and a much earlier counterfactual "ToE (+intercept)") confirms this exclusion as
the cause of late trend-based ToEs — it applies equally to (b). The +intercept column is a
diagnostic, not a coherent test — e(n) is the margin of error on the trend contribution only.

```python
print(f'{"mem":>4} {"slope":>9} {"intercept":>10} {"b·n@end":>9} '
      f'{"int share":>10} {"ToE trend":>10} {"ToE +int":>9}')

shares, deltas = [], []
for i, fit in enumerate(li_fits_all):
    toe_trend, nts, e = li_tsig(fit)
    b, c, n_full = fit['slope'], fit['intercept'], fit['n']
    fitted_end = c + b * n_full                     # fitted separation at F_END

    toe_int = None
    if nts is not None:
        hit = np.where(c + b * nts > e)[0]
        if len(hit):
            toe_int = int(fit['start_year'] + nts[hit[0]] - 1)

    share = c / fitted_end if fitted_end > 0 else np.nan
    shares.append(share)
    if toe_trend is not None and toe_int is not None:
        deltas.append(toe_trend - toe_int)

    print(f'r{i+1:02d}  {b:+9.4f} {c:+10.3f} {b*n_full:9.3f} '
          f'{share:10.1%} {str(toe_trend):>10} {str(toe_int):>9}')

print(f'\nMedian intercept share of fitted separation at {F_END}: '
      f'{np.nanmedian(shares):.1%}')
if deltas:
    print(f'Median ToE delay attributable to excluding the intercept: '
          f'{int(np.median(deltas))} years')
```

```python
from itertools import combinations

# ── Parameter settings (3 sensitivity variants per method) ────────────────────
N_BOOT_RANGE = list(range(3, 10))   # 3..9  (n=10 is exact, no resampling)
ALL_IDX      = list(range(N_MEM))

A_PARAMS = [
    {'alpha': 0.01, 'pct': 99, 'label': 'p=0.01, 99th pct'},
    {'alpha': 0.05, 'pct': 66, 'label': 'p=0.05, 66th pct'},
    {'alpha': 0.10, 'pct': 50, 'label': 'p=0.10, 50th pct'},
]
B_PARAMS = [
    {'alpha': 0.01, 'label': 'p=0.01'},
    {'alpha': 0.05, 'label': 'p=0.05'},
    {'alpha': 0.10, 'label': 'p=0.10'},
]
C_PARAMS = A_PARAMS
D_PARAMS = [
    {'ci': 0.90, 'K': 3,  'label': '90% PI, K=3'},
    {'ci': 0.95, 'K': 5,  'label': '95% PI, K=5'},
    {'ci': 0.99, 'K': 10, 'label': '99% PI, K=10'},
]
E_PARAMS = [
    {'k': 1.0,   'label': '1 σ'},
    {'k': 1.645, 'label': '1.645 σ'},
    {'k': 2.0,   'label': '2 σ'},
]
F_PARAMS = [
    {'ci': 0.90, 'label': '90% (1-sided)'},
    {'ci': 0.95, 'label': '95% (1-sided)'},
    {'ci': 0.99, 'label': '99% (1-sided)'},
]

# ── Method (e) helper (σ and means must be recomputed per subset) ─────────────
def e_toe_from_idx(idx, k, hilla_list=miroc_HiLLA_anom, ssp245_list=miroc_ssp245_anom,
                   noise_win=NOISE_WIN, plot_end=2065):
    noise_years = np.arange(noise_win[0], noise_win[1] + 1)
    ssp_noise = np.stack([ssp245_list[i].sel(time=noise_years).values for i in idx])
    sigma = float((ssp_noise - ssp_noise.mean(axis=0)).std(ddof=1))
    threshold = k * sigma

    common_yrs = np.arange(SAI_START, plot_end + 1)
    for i in idx:
        common_yrs = np.intersect1d(common_yrs, hilla_list[i].time.values.astype(int))
        common_yrs = np.intersect1d(common_yrs, ssp245_list[i].time.values.astype(int))

    h_mean = np.stack([hilla_list[i].sel(time=common_yrs).values for i in idx]).mean(axis=0)
    s_mean = np.stack([ssp245_list[i].sel(time=common_yrs).values for i in idx]).mean(axis=0)
    exceed = np.where((s_mean - h_mean) > threshold)[0]
    return int(common_yrs[exceed[0]]) if len(exceed) else None


def pct_of_valid(toes, pct):
    valid = [t for t in toes if t is not None]
    return int(np.percentile(valid, pct)) if valid else None


# ── Precompute per-pair ToEs (O(N_MEM), not O(C(N_MEM,n))) ────────────────────
print('Precomputing per-pair ToEs...')

a_all_member_toes = {}            # alpha → [toe_pair01, ...]
for p in A_PARAMS:
    a_all_member_toes[p['alpha']] = [
        paired_ttest_toe(h, s, R1_D, alpha=p['alpha'])[0]
        for h, s in zip(miroc_HiLLA_anom, miroc_ssp245_anom)
    ]

b_all_member_toes = {}            # alpha → [...]  (p-values reused from (b) cell)
for p in B_PARAMS:
    b_all_member_toes[p['alpha']] = [tf_toe_from_pvals(pv, alpha=p['alpha'])
                                     for pv in b_pvals_all]

c_all_member_toes = {}            # alpha → [...]
for p in C_PARAMS:
    c_all_member_toes[p['alpha']] = [
        paired_ttest_toe(h, s, R1_D_ENSO, alpha=p['alpha'])[0]
        for h, s in zip(miroc_HiLLA_enso, miroc_ssp245_enso)
    ]

d_all_member_toes = {}            # (ci, K) → [...]
for p in D_PARAMS:
    ci, K = p['ci'], p['K']
    toes = []
    for i in range(N_MEM):
        obs_i = miroc_ssp245_anom[i].sel(time=slice(2020, SAI_START - 1))
        _, lo_i, _ = ar1_forecast(obs_i, PRED_YEARS, ci=ci)
        toes.append(ar1_toe_lower(miroc_HiLLA_anom[i], PRED_YEARS, lo_i, K=K))
    d_all_member_toes[(ci, K)] = toes

f_all_member_toes = {}            # ci → [...]  (fits reused from (f) cell)
for p in F_PARAMS:
    f_all_member_toes[p['ci']] = [li_tsig(fit, ci=p['ci'])[0] for fit in li_fits_all]

print('  Done.')

# ── n=10 exact values ─────────────────────────────────────────────────────────
a_exact = [pct_of_valid(a_all_member_toes[p['alpha']], p['pct']) for p in A_PARAMS]
b_exact = [pct_of_valid(b_all_member_toes[p['alpha']], 50) for p in B_PARAMS]
c_exact = [pct_of_valid(c_all_member_toes[p['alpha']], p['pct']) for p in C_PARAMS]
d_exact = [pct_of_valid(d_all_member_toes[(p['ci'], p['K'])], 50) for p in D_PARAMS]
e_exact = [e_toe_from_idx(ALL_IDX, p['k']) for p in E_PARAMS]
f_exact = [pct_of_valid(f_all_member_toes[p['ci']], 50) for p in F_PARAMS]

print('n=10 exact ToEs:')
for i, p in enumerate(A_PARAMS):  print(f'  (a) {p["label"]:22s}: {a_exact[i]}')
for i, p in enumerate(B_PARAMS):  print(f'  (b) {p["label"]:22s}: {b_exact[i]}')
for i, p in enumerate(C_PARAMS):  print(f'  (c) {p["label"]:22s}: {c_exact[i]}')
for i, p in enumerate(D_PARAMS):  print(f'  (d) {p["label"]:22s}: {d_exact[i]}')
for i, p in enumerate(E_PARAMS):  print(f'  (e) {p["label"]:22s}: {e_exact[i]}')
for i, p in enumerate(F_PARAMS):  print(f'  (f) {p["label"]:22s}: {f_exact[i]}')

# ── Bootstrap (n=3..9): enumerate all C(10,n) subsets ─────────────────────────
print('Running bootstrap (leave-one-out resampling)...')

a_boot = {pi: {n: [] for n in N_BOOT_RANGE} for pi in range(3)}
b_boot = {pi: {n: [] for n in N_BOOT_RANGE} for pi in range(3)}
c_boot = {pi: {n: [] for n in N_BOOT_RANGE} for pi in range(3)}
d_boot = {pi: {n: [] for n in N_BOOT_RANGE} for pi in range(3)}
e_boot = {pi: {n: [] for n in N_BOOT_RANGE} for pi in range(3)}
f_boot = {pi: {n: [] for n in N_BOOT_RANGE} for pi in range(3)}

for n in N_BOOT_RANGE:
    combos = list(combinations(ALL_IDX, n))
    for combo in combos:
        idx = list(combo)
        for pi, p in enumerate(A_PARAMS):
            a_boot[pi][n].append(
                pct_of_valid([a_all_member_toes[p['alpha']][i] for i in idx], p['pct']))
        for pi, p in enumerate(B_PARAMS):
            b_boot[pi][n].append(
                pct_of_valid([b_all_member_toes[p['alpha']][i] for i in idx], 50))
        for pi, p in enumerate(C_PARAMS):
            c_boot[pi][n].append(
                pct_of_valid([c_all_member_toes[p['alpha']][i] for i in idx], p['pct']))
        for pi, p in enumerate(D_PARAMS):
            d_boot[pi][n].append(
                pct_of_valid([d_all_member_toes[(p['ci'], p['K'])][i] for i in idx], 50))
        for pi, p in enumerate(E_PARAMS):
            e_boot[pi][n].append(e_toe_from_idx(idx, p['k']))
        for pi, p in enumerate(F_PARAMS):
            f_boot[pi][n].append(
                pct_of_valid([f_all_member_toes[p['ci']][i] for i in idx], 50))
    print(f'  n={n}: {len(combos)} combinations')

total = sum(len(list(combinations(ALL_IDX, n))) for n in N_BOOT_RANGE)
print(f'Bootstrap complete: {total} subsets per parameter setting.')
```

```python

def plot_ens(ax, arr, yrs, color, light_color, label):
    '''Plot ensemble members (light) and ensemble mean (bold).'''
    for i, row in enumerate(arr):
        ax.plot(yrs, row, color=light_color, lw=0.8, alpha=0.7, label='_')
    ax.plot(yrs, arr.mean(axis=0), color=color, lw=2.0, label=label)


matplotlib.rcParams['font.size'] = 15

HILLA_COL   = '#238B45'
HILLA_LIGHT = '#BAE4B3'
GRAY        = '#636363'
GRAY_LIGHT  = '#D9D9D9'
RED         = '#CB181D'

PARAM_COLORS = ['#1f77b4', '#ff7f0e', '#2ca02c']   # blue, orange, green

GROUP_CENTERS = [0.0, 1.2, 2.4]
BAR_HEIGHT    = 0.40
_N_SENS       = [3, 5, 7, 9, 10]
SENS_COLORS   = {n: plt.cm.managua(i / (len(_N_SENS) - 1)) for i, n in enumerate(_N_SENS)}

def add_sensitivity_panel(ax, boot_data, exact_data, param_labels):
    for n in [3, 5, 7, 9]:
        for pi in range(3):
            vals = [v for v in boot_data[pi][n] if v is not None]
            if len(vals) < 3:
                continue
            p5, p95 = np.percentile(vals, 5), np.percentile(vals, 95)
            ax.barh(GROUP_CENTERS[pi], width=p95 - p5, left=p5,
                    height=BAR_HEIGHT, color=SENS_COLORS[n],
                    edgecolor='none', alpha=1.0,
                    label=f'n={n}' if pi == 0 else '_', zorder=n)
    for pi in range(3):
        if exact_data[pi] is not None:
            y_lo = GROUP_CENTERS[pi] - BAR_HEIGHT / 2 - 0.07
            y_hi = GROUP_CENTERS[pi] + BAR_HEIGHT / 2 + 0.07
            ax.plot([exact_data[pi]] * 2, [y_lo, y_hi],
                    color=SENS_COLORS[10], lw=3.0, zorder=20,
                    label='n=10' if pi == 0 else '_')
    ax.set_yticks(GROUP_CENTERS)
    ax.set_yticklabels(param_labels, fontsize='medium')
    ax.set_ylim(GROUP_CENTERS[0] - 0.65, GROUP_CENTERS[-1] + 0.65)
    ax.grid(ls='--', alpha=0.3, axis='x')
    for sp in ('top', 'right'): ax.spines[sp].set_visible(False)


def cum_count_panel(ax, params, all_toes_dict, key_fn, title, all_yrs):
    for pi, p in enumerate(params):
        member_toes = all_toes_dict[key_fn(p)]
        cum = np.array([sum(1 for t in member_toes if t is not None and t <= y)
                        for y in all_yrs])
        ax.step(all_yrs, cum, color=PARAM_COLORS[pi], lw=2.0, where='post', label=p['label'])
    ax.set_title(title, fontsize='large', pad=6)
    ax.yaxis.set_major_locator(matplotlib.ticker.MaxNLocator(integer=True))
    ax.set_ylim(0, N_MEM + 0.5)
    ax.legend(fontsize='medium', loc='lower right', framealpha=0.85)
    ax.grid(ls='--', alpha=0.3)
    for sp in ('top', 'right'): ax.spines[sp].set_visible(False)


def tidy(ax, legend_loc='upper left'):
    ax.legend(fontsize='medium', loc=legend_loc, framealpha=0.85)
    ax.grid(ls='--', alpha=0.3)
    for sp in ('top', 'right'): ax.spines[sp].set_visible(False)


# ── Figure ────────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(3, 6, figsize=(33, 17), sharex=True)
fig.subplots_adjust(hspace=0.25, wspace=0.30)

all_yrs_f = np.arange(PLOT_START, PLOT_END + 1)

# GMSAT-anomaly panels share a y-scale; difference-series panels share another
for c in (2, 3, 4):
    axes[0, c].sharey(axes[0, 0])
# (b) and (f) are difference-series panels but keep independent y-scales:
# (f)'s trend − e(n) curve dives at small n and would compress (b).

# ══════════════════════════════════════════════════════════════════════════════
# ROW 1 — one illustrative pair / construction per method
# ══════════════════════════════════════════════════════════════════════════════

ax = axes[0, 0]                                             # (a)
ax.plot(miroc_ssp245_anom[0].time.values, miroc_ssp245_anom[0].values,
        color=GRAY, lw=1.8, label='SSP2-4.5 r01')
ax.plot(miroc_HiLLA_anom[0].time.values, miroc_HiLLA_anom[0].values,
        color=HILLA_COL, lw=1.8, label='G6-1.5K-HiLLA r01')
if a_mem_toes[0]:
    ax.axvline(a_mem_toes[0], color=RED, lw=2.0, ls=':', label=f'ToE r01 = {a_mem_toes[0]}')
ax.set_title('(a) paired t-test, same model', fontsize='large', pad=6)
ax.set_ylabel('GMSAT anomaly (°C)', fontsize='medium')
ax.set_ylim(None, 2)
tidy(ax)

ax = axes[0, 1]                                             # (b)
_h0, _cf0 = b_pairs[0]
_cb = np.intersect1d(_h0.time.values.astype(int), _cf0.time.values.astype(int))
_cb = _cb[(_cb >= SAI_START) & (_cb <= B_END)]
db  = _cf0.sel(time=_cb).values - _h0.sel(time=_cb).values
slb, seb, rhob, _, icb = prais_winsten_trend(db)
ax.plot(_cb, db, color=GRAY, lw=1.2, alpha=0.8, label=f'CF − HiLLA (pair 01)')
ax.plot(_cb, icb + slb * np.arange(1, len(_cb) + 1), color='black', lw=1.8,
        label='GLS-AR(1) trend (full window)')
if b_mem_toes[0] is not None:
    ax.axvline(b_mem_toes[0], color=RED, lw=2.0, ls=':', label=f'ToE pair 01 = {b_mem_toes[0]}')
ax.set_title('(b) cross-model GLS trend', fontsize='large', pad=6)
ax.set_ylabel('counterfactual − HiLLA  (K)', fontsize='medium')
tidy(ax)
ax.text(0.02, 0.02, f'[{B_LABEL}]', transform=ax.transAxes,
        fontsize='small', color='#B15928', va='bottom')

ax = axes[0, 2]                                             # (c)
ax.plot(miroc_ssp245_enso[0].time.values, miroc_ssp245_enso[0].values,
        color=GRAY, lw=1.8, label='SSP2-4.5 r01 (ENSO-adj)')
ax.plot(miroc_HiLLA_enso[0].time.values, miroc_HiLLA_enso[0].values,
        color=HILLA_COL, lw=1.8, label='HiLLA r01 (ENSO-adj)')
if c_mem_toes[0]:
    ax.axvline(c_mem_toes[0], color=RED, lw=2.0, ls=':', label=f'ToE r01 = {c_mem_toes[0]}')
ax.set_title('(c) ENSO-adjusted paired t-test', fontsize='large', pad=6)
ax.tick_params(labelleft=False)
tidy(ax)

ax = axes[0, 3]                                             # (d)
ts_s0 = miroc_ssp245_anom[0]
mask_pre = ts_s0.time.values <= SAI_START - 1
ax.plot(ts_s0.time.values[mask_pre], ts_s0.values[mask_pre],
        color=GRAY, lw=1.8, label='SSP2-4.5 r01 (pseudo-obs)')
mu0, lo0, hi0 = ar1_results_all[0]
ax.fill_between(PRED_YEARS, lo0, hi0, color=GRAY_LIGHT, alpha=0.7,
                label='PI (95%, param. unc.)', zorder=1)
ax.plot(PRED_YEARS, mu0, color=GRAY, lw=1.5, ls='--', zorder=2)
ts_h0 = miroc_HiLLA_anom[0]
ax.plot(ts_h0.time.values, ts_h0.values, color=HILLA_COL, lw=1.8,
        label='G6-1.5K-HiLLA r01', zorder=3)
if d_mem_toes[0] is not None:
    ax.axvline(d_mem_toes[0], color=RED, lw=2.0, ls=':',
               label=f'ToE r01 = {d_mem_toes[0]}')
ax.set_title('(d) purely empirical, OLS+AR(1)', fontsize='large', pad=6)
ax.tick_params(labelleft=False)
tidy(ax)

ax = axes[0, 4]                                             # (e)
plot_ens(ax, ssp_arr, ssp_yrs, GRAY, GRAY_LIGHT, 'SSP2-4.5 (n=10)')
plot_ens(ax, hilla_arr, hilla_yrs, HILLA_COL, HILLA_LIGHT, 'G6-1.5K-HiLLA (n=10)')
if sn_toe:
    ax.axvline(sn_toe, color=RED, lw=2.0, ls=':', label=f'ToE = {sn_toe}')
ax.set_title('(e) signal : noise', fontsize='large', pad=6)
ax.tick_params(labelleft=False)
tidy(ax)

ax = axes[0, 5]                                             # (f)
fit0 = li_fits_all[0]
toe0, nts0, e0 = li_tsig(fit0)
yrs_nt  = SAI_START + nts0 - 1
fitline = fit0['intercept'] + fit0['slope'] * nts0
ax.plot(fit0['years'], fit0['d'], color=GRAY, lw=1.2, alpha=0.8,
        label='SSP2-4.5 − HiLLA (r01)')
ax.plot(yrs_nt, fitline, color='black', lw=1.8, label=f'OLS trend (fit to {F_END})')
ax.plot(yrs_nt, fitline - e0, color='black', lw=1.4, ls='--',
        label='trend − e(n)')
ax.axhline(fit0['intercept'], color='k', lw=0.7, alpha=0.25)
if toe0 is not None:
    ax.axvline(toe0, color=RED, lw=2.0, ls=':', label=f'ToE r01 = {toe0}')
ax.set_title('(f) Li et al. analytic $T_{SIG}$', fontsize='large', pad=6)
ax.set_ylabel('SSP2-4.5 − HiLLA  (K)', fontsize='medium')
ax.set_ylim(bottom=float(fit0['d'].min()) - 1.5)   # clip the e(n) dive at small n
tidy(ax)

# ══════════════════════════════════════════════════════════════════════════════
# ROW 2 — detection diagnostics
# ══════════════════════════════════════════════════════════════════════════════

cum_count_panel(axes[1, 0], A_PARAMS, a_all_member_toes, lambda p: p['alpha'], '(g)', all_yrs_f)
axes[1, 0].set_ylabel('# pairs past ToE', fontsize='medium')
cum_count_panel(axes[1, 1], B_PARAMS, b_all_member_toes, lambda p: p['alpha'], '(h)', all_yrs_f)
cum_count_panel(axes[1, 2], C_PARAMS, c_all_member_toes, lambda p: p['alpha'], '(i)', all_yrs_f)
cum_count_panel(axes[1, 3], D_PARAMS, d_all_member_toes,
                lambda p: (p['ci'], p['K']), '(j)', all_yrs_f)

ax = axes[1, 4]                                             # (e): signal vs thresholds
ax.plot(sn_years, sn_signal, color='black', lw=2.0, label='Signal (ens mean diff)')
ax.axhline(0, color='k', lw=0.7, alpha=0.25)
for pi, p in enumerate(E_PARAMS):
    ax.axhline(p['k'] * sn_sigma, color=PARAM_COLORS[pi], lw=1.5, ls='--', label=p['label'])
ax.set_title('(k)', fontsize='large', pad=6)
ax.set_ylabel('Signal = SSP2-4.5 − HiLLA  (K)', fontsize='medium')
tidy(ax)

cum_count_panel(axes[1, 5], F_PARAMS, f_all_member_toes, lambda p: p['ci'], '(l)', all_yrs_f)

# ══════════════════════════════════════════════════════════════════════════════
# ROW 3 — sensitivity to parameters and ensemble size
# ══════════════════════════════════════════════════════════════════════════════

for j, t in enumerate(['(m)', '(n)', '(o)', '(p)', '(q)', '(r)']):
    axes[2, j].set_title(t, fontsize='large', pad=6)

a_labels = [p['label'].replace(', ', ',\n') for p in A_PARAMS]
b_labels = [p['label'] for p in B_PARAMS]
c_labels = [p['label'].replace(', ', ',\n') for p in C_PARAMS]
d_labels = [p['label'].replace(', ', ',\n') for p in D_PARAMS]
e_labels = [p['label'] for p in E_PARAMS]
f_labels = [p['label'].replace(' (', '\n(') for p in F_PARAMS]

add_sensitivity_panel(axes[2, 0],
                      {pi: a_boot[2 - pi] for pi in range(3)},
                      a_exact[::-1], a_labels[::-1])
add_sensitivity_panel(axes[2, 1],
                      {pi: b_boot[2 - pi] for pi in range(3)},
                      b_exact[::-1], b_labels[::-1])
add_sensitivity_panel(axes[2, 2],
                      {pi: c_boot[2 - pi] for pi in range(3)},
                      c_exact[::-1], c_labels[::-1])
add_sensitivity_panel(axes[2, 3], d_boot, d_exact, d_labels)
add_sensitivity_panel(axes[2, 4], e_boot, e_exact, e_labels)
add_sensitivity_panel(axes[2, 5], f_boot, f_exact, f_labels)

axes[0, 0].set_xlim(PLOT_START, 2065)
for ax in axes[2]:
    ax.set_xlabel('Year', fontsize='medium')

from matplotlib.patches import Patch
from matplotlib.lines import Line2D
leg_elements = [
    Patch(facecolor=SENS_COLORS[n], edgecolor='none', label=f'n={n}')
    for n in [3, 5, 7, 9]
] + [Line2D([0], [0], color=SENS_COLORS[10], lw=3.0, label='n=10')]
axes[2, 0].legend(handles=leg_elements, fontsize='medium', loc='lower right', framealpha=0.88)

plt.savefig('Figures/ToE_methods_framework_v4.png', dpi=200, bbox_inches='tight')
plt.show()
```

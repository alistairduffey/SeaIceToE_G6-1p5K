# Time of Emergence — Two Framings (MIROC-ES2H only)
## NH September Sea Ice Area  |  G6-1.5K-HiLLA vs SSP2-4.5

Same two complementary detection methods as v3, but using **MIROC-ES2H 10-member ensembles throughout**.
No historical scenario is used — the SSP2-4.5 record begins in 2020.

- **Perfect-model (left):** MIROC-ES2H 10-member ensembles fully known — Welch's t-test on per-member window means detects when the forced responses first diverge.
- **Obs-only (right):** One SSP2-4.5 member (2020–2034) acts as the single 'observed' record; OLS linear trend + AR(1) residuals extrapolates the counterfactual; ToE when the corresponding HiLLA member exits the 95% prediction interval for 5 consecutive years.

```python
import sys
import fsspec
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import matplotlib
matplotlib.rcParams.update({'font.size': 13})

sys.path.append('../shared/Code_examples/G6-1.5K_example/')
from Utils import set_time_to_center_of_bounds
```

```python
# ── MIROC-ES2H file paths ──────────────────────────────────────────────────────
_MIROC = 's3://reflective-persistent-prod-large/MIROC-ES2H'

miroc_HiLLA_files = [
    f'{_MIROC}/G6-1.5K-HiLLA/Omon/siconc_G6-1.5K-HiLLA_r{i:02d}.nc'
    for i in range(1, 11)
]
miroc_ssp245_files = [
    f'{_MIROC}/G6-1.5K-HiLLA/Omon/siconc_baseline_r{i:02d}.nc'
    for i in range(1, 11)
]
```

```python
def read_miroc_ice(file_list):
    """Load MIROC-ES2H siconc: rename non-standard lat dim; fraction -> %."""
    ds_list = []
    for f in file_list:
        with fsspec.open(f, mode='rb') as fobj:
            ds = xr.open_dataset(fobj, engine='h5netcdf', chunks={}).load()
        extra = [d for d in ds.dims if d not in ('lon', 'time') and d != 'lat']
        if extra:
            ds = ds.rename({extra[0]: 'lat'})
        ice_var = next(
            (v for v in ds.data_vars
             if 'ice' in v.lower() or 'conc' in v.lower() or v == 'SeaIce'), None
        )
        if ice_var and ice_var != 'siconc':
            ds = ds.rename({ice_var: 'siconc'})
        if 'siconc' in ds and float(ds['siconc'].max()) <= 1.5:
            ds['siconc'] = ds['siconc'] * 100.0
        ds_list.append(ds)
    return ds_list


miroc_HiLLA  = read_miroc_ice(miroc_HiLLA_files)
miroc_ssp245 = read_miroc_ice(miroc_ssp245_files)
print(f'MIROC HiLLA:  {len(miroc_HiLLA)} members  | '
      f'{miroc_HiLLA[0].time.values[[0,-1]]}')
print(f'MIROC SSP245: {len(miroc_ssp245)} members  | '
      f'{miroc_ssp245[0].time.values[[0,-1]]}')
```

```python
R_EARTH = 6.371e6  # m


def cell_area_m2(lat, lon):
    """Grid-cell area (m^2) for a regular lat/lon grid."""
    dlat    = np.deg2rad(abs(float(lat[1] - lat[0])))
    dlon    = np.deg2rad(abs(float(lon[1] - lon[0])))
    cos_lat = np.cos(np.deg2rad(lat))
    return (R_EARTH**2 * cos_lat * dlat * dlon).expand_dims({'lon': len(lon)}, axis=1)


def september_sia(ds_list, lat_min=0.0):
    """NH September SIA (10^6 km^2) timeseries, indexed by integer year."""
    result = []
    for ds in ds_list:
        da = ds['siconc']
        lat_nm = 'lat' if 'lat' in da.dims else 'latitude'
        lon_nm = 'lon' if 'lon' in da.dims else 'longitude'
        if float(da[lat_nm][0]) > float(da[lat_nm][-1]):
            da = da.isel(**{lat_nm: slice(None, None, -1)})
        da_nh = da.sel(**{lat_nm: slice(lat_min, 90)})
        area  = cell_area_m2(da_nh[lat_nm], da_nh[lon_nm])
        sia   = ((da_nh / 100.0) * area).sum([lat_nm, lon_nm]) / 1e12
        sep   = sia.sel(time=sia.time.dt.month == 9)
        result.append(sep.assign_coords(time=sep.time.dt.year.values.astype(int)))
    return result


sep_miroc_HiLLA  = september_sia(miroc_HiLLA)
sep_miroc_ssp245 = september_sia(miroc_ssp245)

SAT_START  = 2020   # SSP2-4.5 starts here; no historical available
START_YEAR = 2035   # SAI deployment year

print(f'MIROC HiLLA r01:  years {sep_miroc_HiLLA[0].time.values[[0,-1]]}  '
      f'mean = {float(sep_miroc_HiLLA[0].mean()):.2f} Mkm^2')
print(f'MIROC SSP245 r01: years {sep_miroc_ssp245[0].time.values[[0,-1]]}  '
      f'mean = {float(sep_miroc_ssp245[0].mean()):.2f} Mkm^2')
```

```python
Y_MIN  = 2
Y_MAX  = 40
y_vals = np.arange(Y_MIN, Y_MAX + 1)


def perfect_model_toe(hilla_ts_list, ssp245_ts_list,
                      start_year=START_YEAR, y_vals=y_vals):
    """Welch's one-tailed t-test: per-member window-mean HiLLA > SSP245.

    For each window length Y, compute each member's mean SIA over
    [start_year, start_year+Y-1] and test HiLLA (n=10) > SSP245 (n=10).
    Returns (p_vals, toe_year).
    """
    p_arr = np.full(len(y_vals), np.nan)
    for i, Y in enumerate(y_vals):
        h = [float(ts.sel(time=slice(start_year, start_year + int(Y) - 1)).mean())
             for ts in hilla_ts_list]
        s = [float(ts.sel(time=slice(start_year, start_year + int(Y) - 1)).mean())
             for ts in ssp245_ts_list]
        _, p = stats.ttest_ind(h, s, equal_var=False, alternative='greater')
        p_arr[i] = p
    first = np.where(p_arr < 0.05)[0]
    toe_year = int(start_year + y_vals[first[0]] - 1) if len(first) else None
    return p_arr, toe_year


pm_pvals, pm_toe = perfect_model_toe(sep_miroc_HiLLA, sep_miroc_ssp245)
wlen = (pm_toe - START_YEAR + 1) if pm_toe else None
print(f'Perfect-model ToE: {pm_toe}  (window = {wlen} yr)')
```

```python
PRED_YEARS   = np.arange(START_YEAR, 2085)
K_CONSEC     = 5
CI_LEVEL     = 0.95
TREND_START  = SAT_START              # fit OLS over full pre-SAI SSP245 period
NOISE_PERIOD = (SAT_START, START_YEAR - 1)  # 2020-2034


def ar1_forecast(obs_ts, pred_years, trend_start=TREND_START, ci=CI_LEVEL):
    """OLS linear trend + AR(1) residuals h-step forecast.

    Steps:
      1. Fit OLS line to obs_ts from trend_start onwards.
      2. Compute residuals; fit AR(1):  e_t = phi*e_{t-1} + eta,  eta ~ N(0, sigma^2).
      3. h-step mean:  trend(t) + phi^h * e_last.
      4. h-step PI:    mean +/- z*sigma*sqrt((1 - phi^{2h}) / (1 - phi^2)).

    Returns (mu, lower_ci, upper_ci).
    """
    mask  = obs_ts.time.values >= trend_start
    x_fit = obs_ts.time.values[mask].astype(float)
    y_fit = obs_ts.values[mask].astype(float)

    slope, intercept, *_ = stats.linregress(x_fit, y_fit)

    resid    = y_fit - (slope * x_fit + intercept)
    phi      = np.dot(resid[1:], resid[:-1]) / np.dot(resid[:-1], resid[:-1])
    innov    = resid[1:] - phi * resid[:-1]
    sigma    = np.sqrt(np.sum(innov**2) / (len(innov) - 1))
    eps_last = resid[-1]

    h   = pred_years.astype(float) - x_fit[-1]
    mu  = slope * pred_years + intercept + phi**h * eps_last

    z       = stats.norm.ppf((1 + ci) / 2)
    pi_half = z * sigma * np.sqrt((1 - phi**(2*h)) / max(1 - phi**2, 1e-8))

    print(f'OLS slope:  {slope:.4f} Mkm^2/yr  ({int(x_fit[0])}-{int(x_fit[-1])})')
    print(f'AR(1):  phi = {phi:.3f},  sigma_eta = {sigma:.3f} Mkm^2')
    return mu, mu - pi_half, mu + pi_half


def ar1_toe(hilla_ts, pred_years, upper, start_year=START_YEAR, K=K_CONSEC):
    """First year where HiLLA > upper PI for K consecutive years."""
    h_yrs  = hilla_ts.time.values
    common = np.intersect1d(h_yrs[h_yrs >= start_year], pred_years)
    if len(common) < K:
        return None
    h_vals = hilla_ts.sel(time=common).values
    ui     = upper[np.searchsorted(pred_years, common)]
    above  = h_vals > ui
    for i in range(len(above) - K + 1):
        if above[i : i + K].all():
            return int(common[i])
    return None


# r01 as the 'observed' baseline: pre-SAI SSP245 period only (2020-2034)
obs_r01 = sep_miroc_ssp245[0].sel(time=slice(SAT_START, START_YEAR - 1))
ar1_mu, ar1_lower, ar1_upper = ar1_forecast(obs_r01, PRED_YEARS)
obs_toe = ar1_toe(sep_miroc_HiLLA[0], PRED_YEARS, ar1_upper)
print(f'Obs-only ToE (r01, K={K_CONSEC}): {obs_toe}')
```

```python
BLUE       = '#2171B5'
BLUE_LIGHT = '#BDD7E7'
GRAY       = '#636363'
GRAY_LIGHT = '#D9D9D9'


def ts_stack(ts_list):
    """Stack year-indexed 1-D DataArrays into (n_members, n_years); common years only."""
    common = sorted(set.intersection(*[set(ts.time.values.tolist()) for ts in ts_list]))
    return np.stack([ts.sel(time=common).values for ts in ts_list]), np.array(common, dtype=int)


def plot_ens(ax, arr, years, color, light, label):
    mean = arr.mean(axis=0)
    p05  = np.percentile(arr,  5, axis=0)
    p95  = np.percentile(arr, 95, axis=0)
    ax.fill_between(years, p05, p95, color=light, alpha=0.55, zorder=1)
    ax.plot(years, mean, color=color, lw=2.2, label=label, zorder=2)


fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5.5))
fig.subplots_adjust(wspace=0.04)

# ── Left panel: perfect-model ──────────────────────────────────────────────────
hilla_arr, hilla_years = ts_stack(sep_miroc_HiLLA)
ssp_arr,   ssp_years   = ts_stack(sep_miroc_ssp245)

plot_ens(ax1, ssp_arr,   ssp_years,   GRAY, GRAY_LIGHT, 'SSP2-4.5  (n=10)')
plot_ens(ax1, hilla_arr, hilla_years, BLUE, BLUE_LIGHT, 'G6-1.5K-HiLLA  (n=10)')

ax1.axvline(START_YEAR, color='k', lw=1.2, ls='--', alpha=0.45, label='SAI start (2035)')

if pm_toe:
    ax1.axvline(pm_toe, color=BLUE, lw=2, ls=':', zorder=5)
    ax1.text(pm_toe + 0.5, 0.96, f'ToE = {pm_toe}',
             transform=ax1.get_xaxis_transform(),
             color=BLUE, fontsize=11, va='top', fontweight='bold')

ax1.set_xlim(SAT_START - 2, 2080)
ax1.set_xlabel('Year')
ax1.set_ylabel('NH September SIA  (10\u2076 km\u00b2)')
ax1.set_title('Perfect-model framing  (MIROC-ES2H)\n'
              'Known 10-member ensembles \u2014 two-sample test', fontsize=11, pad=8)
ax1.legend(fontsize=10, loc='upper right', framealpha=0.9)
ax1.grid(ls='--', alpha=0.3)
for sp in ('top', 'right'):
    ax1.spines[sp].set_visible(False)

# ── Right panel: obs-only (r01 as 'observed') ──────────────────────────────────
_obs_plot = sep_miroc_ssp245[0].sel(time=slice(SAT_START, START_YEAR - 1))
ax2.plot(_obs_plot.time.values, _obs_plot.values,
         color=GRAY, lw=2.2, label='"Observed"  (SSP2-4.5 r01, 2020\u20132034)')

ax2.fill_between(PRED_YEARS, ar1_lower, ar1_upper,
                 color=GRAY_LIGHT, alpha=0.7, zorder=1,
                 label=f'AR(1) counterfactual  ({int(CI_LEVEL*100)}% PI)')
ax2.plot(PRED_YEARS, ar1_mu, color=GRAY, lw=1.8, ls='--', zorder=2)

h_yrs = sep_miroc_HiLLA[0].time.values[sep_miroc_HiLLA[0].time.values <= 2080]
ax2.plot(h_yrs, sep_miroc_HiLLA[0].sel(time=h_yrs).values,
         color=BLUE, lw=2.2, label='G6-1.5K-HiLLA  r01', zorder=3)

ax2.axvline(START_YEAR, color='k', lw=1.2, ls='--', alpha=0.45, label='SAI start  (2035)')

if obs_toe:
    ax2.axvline(obs_toe, color=BLUE, lw=2, ls=':', zorder=5)
    ax2.text(obs_toe + 0.5, 0.96, f'ToE = {obs_toe}\n(K={K_CONSEC} yr)',
             transform=ax2.get_xaxis_transform(),
             color=BLUE, fontsize=11, va='top', fontweight='bold')

ax2.set_xlim(SAT_START - 2, 2080)
ax2.set_xlabel('Year')
ax2.set_ylabel('NH September SIA  (10\u2076 km\u00b2)')
ax2.set_title('Obs-only framing  (MIROC-ES2H r01)\n'
              'Single realisation \u2014 linear trend + AR(1) counterfactual', fontsize=11, pad=8)
ax2.legend(fontsize=10, loc='lower left', framealpha=0.9)
ax2.grid(ls='--', alpha=0.3)
for ax in ax1, ax2:
    ax.set_ylim(0, None)
for sp in ('top', 'right'):
    ax2.spines[sp].set_visible(False)

fig.suptitle(
    'Time of Emergence  |  NH September Sea Ice Area  |  G6-1.5K-HiLLA vs SSP2-4.5'
    '  (MIROC-ES2H)',
    fontsize=13, y=1.02
)
plt.tight_layout()
plt.savefig('Figures/ToE_two_panel_v5.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
SNR_THRESHOLD = 2.0
SNR_ROLL_K    = K_CONSEC


def snr_toe(hilla_ts_list, ssp245_ts_list,
            start_year=START_YEAR, threshold=SNR_THRESHOLD,
            noise_period=NOISE_PERIOD):
    """Signal-to-noise ToE (perfect-model or multi-member case).

    Noise = std of pooled (member - ensemble-mean) residuals over noise_period.
    Signal = instantaneous difference of ensemble means.
    Returns first year where signal > threshold * noise, or None.
    """
    noise_arr = np.stack([
        ts.sel(time=slice(*noise_period)).values for ts in ssp245_ts_list
    ])
    ensmean   = noise_arr.mean(axis=0)
    residuals = (noise_arr - ensmean).ravel()
    noise     = float(np.std(residuals, ddof=1))

    h_yrs = sorted(set.intersection(*[set(ts.time.values.tolist()) for ts in hilla_ts_list]))
    s_yrs = sorted(set.intersection(*[set(ts.time.values.tolist()) for ts in ssp245_ts_list]))
    years = np.array(sorted(set(h_yrs) & set(s_yrs)))
    years = years[years >= start_year]
    if len(years) == 0:
        return None

    h_arr  = np.stack([ts.sel(time=years).values for ts in hilla_ts_list])
    s_arr  = np.stack([ts.sel(time=years).values for ts in ssp245_ts_list])
    signal = h_arr.mean(axis=0) - s_arr.mean(axis=0)

    above = np.where(signal > threshold * noise)[0]
    return int(years[above[0]]) if len(above) else None


def loo_toes(hilla_list, ssp245_list, toe_func):
    """Leave-one-out jackknife: ToE for each (n-1)-member subset."""
    n = len(hilla_list)
    return [
        toe_func(
            [ts for i, ts in enumerate(hilla_list)  if i != k],
            [ts for i, ts in enumerate(ssp245_list) if i != k],
        )
        for k in range(n)
    ]


pm_snr_toe = snr_toe(sep_miroc_HiLLA, sep_miroc_ssp245)

loo_ttest = loo_toes(sep_miroc_HiLLA, sep_miroc_ssp245,
                     lambda h, s: perfect_model_toe(h, s)[1])
loo_snr   = loo_toes(sep_miroc_HiLLA, sep_miroc_ssp245, snr_toe)

def _fmt(toes):
    valid = [t for t in toes if t is not None]
    return f'{min(valid)}-{max(valid)}' if valid else 'never detected'

print(f'PM t-test:  central={pm_toe},    LOO range={_fmt(loo_ttest)}')
print(f'PM S/N:     central={pm_snr_toe}, LOO range={_fmt(loo_snr)}')
```

```python
def obs_snr_toe_rolling(ssp245_ts, hilla_ts,
                        start_year=START_YEAR, threshold=SNR_THRESHOLD,
                        K=SNR_ROLL_K, noise_period=NOISE_PERIOD):
    """Obs-only S/N ToE with K-year trailing rolling mean.

    Noise = std of K-year rolling means of OLS-detrended SSP245 over noise_period.
    Signal = K-year trailing rolling mean of (HiLLA - SSP245).
    ToE = first year Y where rolling_mean_signal > threshold * noise_smooth.
    """
    ssp_base = ssp245_ts.sel(time=slice(*noise_period)).values
    n_base   = len(ssp_base)
    x_base   = np.arange(n_base, dtype=float)
    _slope, _intercept, *_ = stats.linregress(x_base, ssp_base)
    ssp_detrended = ssp_base - (_slope * x_base + _intercept)
    rolling_base  = np.array([ssp_detrended[j : j+K].mean() for j in range(n_base - K + 1)])
    noise_smooth  = float(np.std(rolling_base, ddof=1))

    h_yrs  = hilla_ts.time.values
    s_yrs  = ssp245_ts.time.values
    common = np.array(sorted(
        set(h_yrs[h_yrs >= start_year].tolist()) & set(s_yrs.tolist())
    ))
    if len(common) < K:
        return None, noise_smooth

    diff = hilla_ts.sel(time=common).values - ssp245_ts.sel(time=common).values

    rolling_sig = np.array([diff[j : j+K].mean() for j in range(len(diff) - K + 1)])
    rolling_yrs = np.array([common[j + K - 1]    for j in range(len(diff) - K + 1)])

    above = np.where(rolling_sig > threshold * noise_smooth)[0]
    toe   = int(rolling_yrs[above[0]]) if len(above) else None
    return toe, noise_smooth


# Per-member: each SSP245 member (2020-2034) acts as 'obs'; test paired HiLLA member
obs_ar1_toes = []
obs_snr_toes = []

for i, (ssp245_i, hilla_i) in enumerate(
        zip(sep_miroc_ssp245, sep_miroc_HiLLA), start=1):

    # AR(1) counterfactual trained on pre-SAI SSP245 member
    obs_i = ssp245_i.sel(time=slice(SAT_START, START_YEAR - 1))
    _, _, up_i = ar1_forecast(obs_i, PRED_YEARS)
    toe_ar1_i  = ar1_toe(hilla_i, PRED_YEARS, up_i)
    obs_ar1_toes.append(toe_ar1_i)

    # Rolling S/N
    toe_snr_i, noise_i = obs_snr_toe_rolling(ssp245_i, hilla_i)
    obs_snr_toes.append(toe_snr_i)

    print(f'r{i:02d}: AR(1) ToE={toe_ar1_i},  '
          f'S/N rolling ToE={toe_snr_i}  '
          f'(noise_smooth={noise_i:.3f} Mkm^2, K={SNR_ROLL_K})')
```

```python
def _toe_stats(toes):
    """(min, max, valid_list) of non-None ToE values."""
    valid = [t for t in toes if t is not None]
    return (min(valid), max(valid), valid) if valid else (None, None, [])


fig2, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 3.2))
fig2.subplots_adjust(wspace=0.35)

BAR_H = 0.45


def draw_method_bar(ax, y_pos, toes, central_toe, col, light):
    """One horizontal bar: shaded range + individual ticks + diamond for central."""
    lo, hi, valid = _toe_stats(toes)
    if lo is not None and hi != lo:
        ax.barh(y_pos, hi - lo, left=lo, height=BAR_H,
                color=light, alpha=0.85, zorder=1, linewidth=0)
    if valid:
        ax.vlines(valid, y_pos - BAR_H/2, y_pos + BAR_H/2,
                  color=col, lw=1.8, zorder=2)
    if central_toe is not None:
        ax.scatter([central_toe], [y_pos], color=col, s=90,
                   zorder=4, marker='D')
    elif valid:
        ax.scatter([np.median(valid)], [y_pos], color=col, s=90,
                   zorder=4, marker='D')


# ── Left: perfect-model LOO ────────────────────────────────────────────────────
draw_method_bar(ax1, 2, loo_ttest, pm_toe,     BLUE, BLUE_LIGHT)
draw_method_bar(ax1, 1, loo_snr,   pm_snr_toe, GRAY, GRAY_LIGHT)

ax1.set_xlim(2035, 2080)
ax1.set_ylim(0.4, 2.8)
ax1.set_yticks([1, 2])
ax1.set_yticklabels(['S/N  (2\u03c3)', 't-test'], fontsize=11)
ax1.set_xlabel('Year of Emergence')
ax1.set_title('Perfect-model  (MIROC-ES2H) LOO \u2014 10 members', fontsize=11, pad=8)
ax1.grid(axis='x', ls='--', alpha=0.3)
for sp in ('top', 'right'):
    ax1.spines[sp].set_visible(False)

# ── Right: obs-only (10 paired MIROC members) ──────────────────────────────────
draw_method_bar(ax2, 2, obs_ar1_toes, obs_ar1_toes[0], BLUE, BLUE_LIGHT)
draw_method_bar(ax2, 1, obs_snr_toes, obs_snr_toes[0], GRAY, GRAY_LIGHT)

ax2.set_xlim(2035, 2080)
ax2.set_ylim(0.4, 2.8)
ax2.set_yticks([1, 2])
ax2.set_yticklabels([f'S/N rolling\n(2\u03c3, K={SNR_ROLL_K})', 'AR(1)+OLS'], fontsize=11)
ax2.set_xlabel('Year of Emergence')
ax2.set_title('Obs-only  (MIROC-ES2H) 10 paired members', fontsize=11, pad=8)
ax2.grid(axis='x', ls='--', alpha=0.3)
for sp in ('top', 'right'):
    ax2.spines[sp].set_visible(False)

from matplotlib.lines import Line2D
from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor=BLUE_LIGHT, label='Range (LOO / members)'),
    Line2D([0], [0], color=BLUE, lw=0, marker='D', markersize=8,
           label='Central / r01 estimate'),
    Line2D([0], [0], color=GRAY, lw=1.5, label='Individual estimates'),
]
fig2.legend(handles=legend_elements, loc='lower center', ncol=3,
            fontsize=10, framealpha=0.9, bbox_to_anchor=(0.5, -0.18))

fig2.suptitle(
    'ToE sensitivity to method  |  NH September SIA  |  G6-1.5K-HiLLA vs SSP2-4.5'
    '  (MIROC-ES2H)',
    fontsize=12, y=1.04
)
plt.tight_layout()
plt.savefig('Figures/ToE_uncertainty_v5.png', dpi=200, bbox_inches='tight')
plt.show()
```
